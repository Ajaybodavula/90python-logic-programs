```python
# 9. Feature matrix X (5 samples, 3 features)
import numpy as np
np.random.seed(42)
X = np.random.randint(10, 100, size=(5, 3))
print("9 Shape:", X.shape)
```

```python
# 10. Mean of each feature across samples (axis=0)
import numpy as np
feature_means = X.mean(axis=0)
print("10 Feature Means:", feature_means)
```

```python
# 11. Filter rows where first feature > 50
import numpy as np
filtered_X = X[X[:, 0] > 50]
print("11 Filtered Rows:\n", filtered_X)
```

```python
# 12. Standardize every feature column (Z-score normalization)
import numpy as np
means = X.mean(axis=0)
stds = X.std(axis=0)
X_standardized = (X - means) / stds
print("12 Standardized X:\n", X_standardized)
```

```python
# 13. Random matrix (10, 4) with fixed seed
import numpy as np
np.random.seed(123)
rand_matrix = np.random.rand(10, 4)
print("13 Shape:", rand_matrix.shape)
```

```python
# 14. Matrix-vector multiplication (X @ weights)
import numpy as np
weights = np.array([0.5, 0.2, 0.8])
predictions = X @ weights
print("14 Matrix Multiplication Output:\n", predictions)
```

```python
# 15. Add bias vector using broadcasting
import numpy as np
bias = np.array([1.0, 0.5, -0.5])
X_biased = X + bias
print("15 Broadcasted Addition:\n", X_biased)
```

```python
# 16. Mean Squared Error (MSE)
import numpy as np
y_true = np.array([2.5, 0.0, 2.1, 7.8])
y_pred = np.array([3.0, -0.5, 2.0, 7.2])
mse = np.mean((y_true - y_pred) ** 2)
print("16 MSE:", mse)
```

```python
#20. Explain the difference between `A * B` and `A @ B` in your own words. for this 
A * B (Element-wise Multiplication)
Multiplies corresponding elements at the same position in both arrays item-by-item.
Both arrays must have identical or broadcastable shapes (e.g., [1, 2] * [3, 4] = [3, 8]).
A @ B (Matrix Multiplication)
Computes the standard linear algebra dot product between rows of A and columns of B. 
Requires the number of columns in A to match the number of rows in B (e.g..,(m * n) @ (n * p) --> (m * p))
```
