```python
##Q79. Read numbers from file and write even, odd and prime numbers to separate files
with open("numbers.txt", "w") as file:
    file.write("10 15 3 7 8 12 2 1 0 -5 13")

with open("numbers.txt", "r") as file:
    numbers = list(map(int, file.read().split()))

with open("even.txt", "w") as even_file, \
     open("odd.txt", "w") as odd_file, \
     open("prime.txt", "w") as prime_file:

    for n in numbers:
        if n % 2 == 0:
            even_file.write(f"{n} ")
        else:
            odd_file.write(f"{n} ")

        if n > 1:
            count = 0
            for i in range(1, n + 1):
                if n % i == 0:
                    count += 1
            if count == 2:
                prime_file.write(f"{n} ")

print("Files created")
```

```python
##Q83. Check whether a number is Prime, Armstrong or Perfect using functions
def prime(n):
    if n <= 1:
        return False

    count = 0

    for i in range(1, n + 1):
        if n % i == 0:
            count += 1

    return count == 2


def armstrong(n):
    original = n
    digits = len(str(n))
    total = 0

    while n > 0:
        digit = n % 10
        total += digit ** digits
        n //= 10

    return total == original


def perfect(n):
    total = 0

    for i in range(1, n):
        if n % i == 0:
            total += i

    return total == n


n = int(input("Enter a number: "))

if prime(n):
    print("Prime number")
else:
    print("Not a prime number")

if armstrong(n):
    print("Armstrong number")
else:
    print("Not an Armstrong number")

if perfect(n):
    print("Perfect number")
else:
    print("Not a perfect number")
```

```python
##Q85. Print all Strong numbers between given interval using functions
def strong(n):
    original = n
    total = 0

    while n > 0:
        digit = n % 10

        factorial = 1

        for i in range(1, digit + 1):
            factorial *= i

        total += factorial
        n //= 10

    return total == original


start = int(input("Enter starting number: "))
end = int(input("Enter ending number: "))

for n in range(start, end + 1):
    if strong(n):
        print(n)
```

```python
##Q89. Display the ATM Transaction
balance = 10000

print("1. Check Balance")
print("2. Deposit Money")
print("3. Withdraw Money")

choice = int(input("Enter your choice: "))

if choice == 1:
    print("Balance =", balance)

elif choice == 2:
    amount = int(input("Enter amount to deposit: "))
    balance += amount
    print("Balance =", balance)

elif choice == 3:
    amount = int(input("Enter amount to withdraw: "))

    if amount <= balance:
        balance -= amount
        print("Please collect your cash")
        print("Balance =", balance)
    else:
        print("Insufficient balance")

else:
    print("Invalid choice")
```
